```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def evaluate_model(y_true, y_pred):
    """Calculate common classification metrics."""
    results = {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, average="weighted", zero_division=0),
        "Recall": recall_score(y_true, y_pred, average="weighted", zero_division=0),
        "F1 Score": f1_score(y_true, y_pred, average="weighted", zero_division=0)
    }
    return results

print("Common evaluation function created successfully.")
```

```python
def predict_model(model, X_test):
    """Generate predictions using the common prediction pipeline."""
    predictions = model.predict(X_test)
    return predictions

print("Common prediction function created successfully.")
```

```python
def train_model(model, X_train, y_train):
    """Train a model using the common training pipeline."""
    model.fit(X_train, y_train)
    return model

print("Common training function created successfully.")
```
